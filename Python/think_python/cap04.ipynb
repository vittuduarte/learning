```python
from jupyturtle import penup, pendown, forward, left, right, make_turtle
```

Escreva uma função chamada rectangle que desenhe um retangulo com os comprimentos dos lados fornecidos.

Escreva uma função chamada rhombus que desenhe um losango com um comprimento de lado e um angulo interno fornecidos.

```python
def rectangle(t, width, height):
    for i in range(2):
        forward(width)
        left(90)
        forward(height)
        left(90)
        
bob = make_turtle(delay=0.1)
rectangle(bob, 100, 50)
```

```python
bob = make_turtle(delay=0.1)
def rhombus(t, size):
    for i in range(2):
        forward(size)
        left(60)
        forward(size)
        left(120)

rhombus(bob, 100)
```
